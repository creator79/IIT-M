## Solutions

```
# Answers 1?30 use reusable traversal functions for the university dictionary.
def projects(system):
    for university, udata in system['universities'].items():
        for faculty, fdata in udata['faculties'].items():
            for department, ddata in fdata['departments'].items():
                for lab, ldata in ddata.get('labs',{}).items():
                    for project, pdata in ldata.get('research_projects',{}).items():
                        yield university, faculty, department, lab, project, pdata
print('Q1:',university_system['universities']['GTU_India']['location'])
print('Q4:',list(university_system['universities']['GTU_India']['faculties']['engineering']['departments']['computer_science']['labs']))
print('Q6:',[p[4] for p in projects(university_system)])
print('Q7:',[p[4] for p in projects(university_system) if p[5]['status']=='ongoing'])
print('Q15:',[p[4] for p in projects(university_system)])
print('Q21:',max(projects(university_system),key=lambda p:p[5]['funding'])[4])
print('Q22:',sum(p[5]['funding'] for p in projects(university_system)))
print('Q23:',min(university_system['universities'],key=lambda u:university_system['universities'][u]['ranking']))
print('Q24:',[p[4] for p in projects(university_system) if p[5]['status']=='completed'])
print('Q30 helper: next((p for p in projects(university_system) if p[4] == project_name), None)')
```
