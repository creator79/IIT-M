## Solutions

````python
# Questions 113: dictionary checks, searches, missing values, and updates
print(102 in employees, 110 in employees)
print('Alice' in marks)
print('Monitor' in cart)
print('Grade A:',[name for name,grade in grades.items() if grade=='A'])
print('Price 80:',[item for item,price in prices.items() if price==80])
print('Score 78:',[name for name,score in runs.items() if score==78])
print('Missing phones:',[name for name,phone in customers.items() if phone is None])
print('Missing emails:',[name for name,email in emails.items() if email is None]); emails={k:(v or 'Not Available') for k,v in emails.items()}
print('Out of stock:',[title for title,count in books.items() if count==0]); print('Incomplete patients:',[name for name,blood in patients.items() if not blood])
print('Missing vehicle data:',[reg for reg,kind in vehicles.items() if kind is None]); print('Unrecorded attendance:',[name for name,state in attendance.items() if state is None]); print('Unrated movies:',[name for name,rating in ratings.items() if rating is None])
````
