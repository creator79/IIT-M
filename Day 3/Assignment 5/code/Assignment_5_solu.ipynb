```python
Q1.Write a program to check whether Employee ID 1050 exists.
Q2.Ask the user to enter an Employee ID and display whether it exists.
Q3.If the Employee ID exists, display all the employee details.
Q4.If the Employee ID does not exist, display:Employee not found.
Q5.Display the salary of Employee 1032.
Q6.Display the department of Employee 1087.
Q7.Print only the email address of Employee 1015.
Q8.Find all employees whose email address is missing (None).
Q9.Find all employees whose phone number is missing.
Q10.Count how many employees have incomplete records (missing email or phone).
Q11.Replace all missing email addresses with: Not Available
Q12.Find all employees working in the IT department.
Q13.Find all employees working in the Finance department.
Q14.Display the names of all employees managed by Manager_4.
Q15.Find all employees older than 50 years.
Q16.Find all employees younger than 30 years.
Q17.Find the employee with the highest salary.
Q18.Find the employee with the lowest salary.
Q19.Calculate the average salary of all employees.
Q20.Display employees earning more than ₹100,000.
Q21.Display employees earning less than ₹50,000.
Q22.Count how many employees belong to each department.

Example output

HR : 18
Finance : 15
IT : 22
Sales : 17
Marketing : 14
Operations : 14

Q23.Count how many employees are managed by each manager.
Q24. Find which department has the largest number of employees.
Q25.Increase the salary of every employee by 10%.
Q26.Update the department of Employee 1040 to Research.
Q27.Add a new employee to the dictionary.
Q28.Delete Employee 1018 from the dictionary.
Q29.Update the phone number of Employee 1075.
```

```python
# Answers Q1-Q29

import json
from collections import Counter

with open("employees.json", "r") as file:
    employees = json.load(file)

# Q1
print("Q1:", "1050" in employees)

# Q2
emp_id = input("Enter Employee ID: ")
print(
    "Employee exists"
    if str(emp_id) in employees
    else "Employee not found"
)

# Q3-Q4
emp_id = input("Enter Employee ID: ")
if str(emp_id) in employees:
    print(employees[str(emp_id)])
else:
    print("Employee not found.")

# Q5
print("Q5 Salary:", employees["1032"]["Salary"])

# Q6
print("Q6 Department:", employees["1087"]["Department"])

# Q7
print("Q7 Email:", employees["1015"]["Email"])

# Q8
print(
    "Q8 Missing email:",
    [e for e, d in employees.items() if d.get("Email") is None]
)

# Q9
print(
    "Q9 Missing phone:",
    [e for e, d in employees.items() if d.get("Phone") is None]
)

# Q10
print(
    "Q10 Incomplete records:",
    sum(
        d.get("Email") is None or d.get("Phone") is None
        for d in employees.values()
    )
)

# Q11
for d in employees.values():
    if d.get("Email") is None:
        d["Email"] = "Not Available"

print("Q11 Missing emails replaced with 'Not Available'")

# Q12
print(
    "Q12 IT:",
    [e for e, d in employees.items()
     if d.get("Department") == "IT"]
)

# Q13
print(
    "Q13 Finance:",
    [e for e, d in employees.items()
     if d.get("Department") == "Finance"]
)

# Q14
print(
    "Q14 Manager_4:",
    [d.get("Name") for d in employees.values()
     if d.get("Manager") == "Manager_4"]
)

# Q15
print(
    "Q15 Older than 50:",
    [d.get("Name") for d in employees.values()
     if d.get("Age", 0) > 50]
)

# Q16
print(
    "Q16 Younger than 30:",
    [d.get("Name") for d in employees.values()
     if d.get("Age", 0) < 30]
)

# Q17
highest = max(
    employees.items(),
    key=lambda item: item[1].get("Salary", 0)
)
print("Q17 Highest salary:", highest)

# Q18
lowest = min(
    employees.items(),
    key=lambda item: item[1].get("Salary", 0)
)
print("Q18 Lowest salary:", lowest)

# Q19
average_salary = (
    sum(d.get("Salary", 0) for d in employees.values())
    / len(employees)
)
print("Q19 Average salary:", average_salary)

# Q20
print(
    "Q20 Salary > 100000:",
    [d.get("Name") for d in employees.values()
     if d.get("Salary", 0) > 100000]
)

# Q21
print(
    "Q21 Salary < 50000:",
    [d.get("Name") for d in employees.values()
     if d.get("Salary", 0) < 50000]
)

# Q22
department_count = Counter(
    d.get("Department") for d in employees.values()
)
print("Q22 Department counts:", dict(department_count))

# Q23
manager_count = Counter(
    d.get("Manager") for d in employees.values()
)
print("Q23 Manager counts:", dict(manager_count))

# Q24
largest_department = max(
    department_count,
    key=department_count.get
)
print("Q24 Largest department:", largest_department)

# Q25
for d in employees.values():
    d["Salary"] = d.get("Salary", 0) * 1.10

print("Q25 Salary increased by 10%")

# Q26
employees["1040"]["Department"] = "Research"
print(
    "Q26 Updated department:",
    employees["1040"]["Department"]
)

# Q27
employees["1100"] = {
    "Name": "New Employee",
    "Age": 25,
    "Department": "IT",
    "Salary": 60000,
    "Manager": "Manager_1",
    "Email": "newemployee@company.com",
    "Phone": "9999999999"
}

print("Q27 New employee:", employees["1100"])

# Q28
removed_employee = employees.pop("1018", None)
print("Q28 Removed employee:", removed_employee)

# Q29
employees["1075"]["Phone"] = "9999999999"
print(
    "Q29 Updated phone:",
    employees["1075"]["Phone"]
)
```

```python
# Answers Q30-Q40

# Q30 Generate summary report
salaries = [d.get("Salary", 0) for d in employees.values()]

print("Q30 Summary Report")
print("Total Employees:", len(employees))
print(
    "Employees with Missing Email:",
    sum(d.get("Email") is None for d in employees.values())
)
print(
    "Employees with Missing Phone:",
    sum(d.get("Phone") is None for d in employees.values())
)
print("Average Salary:", sum(salaries) / len(salaries))
print("Highest Salary:", max(salaries))
print("Lowest Salary:", min(salaries))


# Q31 Department-wise report
department_employees = {}

for d in employees.values():
    department = d.get("Department")
    name = d.get("Name")
    department_employees.setdefault(department, []).append(name)

print("\nQ31 Department-wise Report")

for department, names in department_employees.items():
    print("\n" + str(department))
    print("-" * len(str(department)))

    for name in names:
        print(name)


# Q32 Employees sorted by salary (highest to lowest)
sorted_employees = sorted(
    employees.items(),
    key=lambda item: item[1].get("Salary", 0),
    reverse=True
)

print("\nQ32 Employees Sorted by Salary:")

for emp_id, d in sorted_employees:
    print(
        emp_id,
        d.get("Name"),
        d.get("Salary")
    )


# Q33 Names containing "25"
print(
    "Q33:",
    [
        d.get("Name")
        for d in employees.values()
        if "25" in str(d.get("Name", ""))
    ]
)


# Q34 Salary between 60000 and 80000
print(
    "Q34:",
    [
        d.get("Name")
        for d in employees.values()
        if 60000 <= d.get("Salary", 0) <= 80000
    ]
)


# Q35 Managers supervising more than 8 employees
manager_count = Counter(
    d.get("Manager")
    for d in employees.values()
)

print(
    "Q35:",
    {
        manager: count
        for manager, count in manager_count.items()
        if count > 8
    }
)


# Q36 Department with highest average salary
dept_salaries = {}

for d in employees.values():
    department = d.get("Department")
    salary = d.get("Salary", 0)

    dept_salaries.setdefault(department, []).append(salary)

dept_avg = {
    department: sum(salaries) / len(salaries)
    for department, salaries in dept_salaries.items()
}

highest_avg_department = max(
    dept_avg,
    key=dept_avg.get
)

print("Q36:", highest_avg_department)


# Q37 Both email and phone missing
print(
    "Q37:",
    [
        d.get("Name")
        for d in employees.values()
        if d.get("Email") is None
        and d.get("Phone") is None
    ]
)


# Q38 Duplicate phone numbers
phone_groups = {}

for emp_id, d in employees.items():
    phone = d.get("Phone")

    if phone is not None:
        phone_groups.setdefault(phone, []).append(emp_id)

duplicates = {
    phone: ids
    for phone, ids in phone_groups.items()
    if len(ids) > 1
}

print("Q38:", duplicates)


# Q39 Email domain company.com
print(
    "Q39:",
    [
        d.get("Name")
        for d in employees.values()
        if isinstance(d.get("Email"), str)
        and d.get("Email").endswith("@company.com")
    ]
)


# Q40 New dictionary containing only IT employees
it_employees = {
    emp_id: d
    for emp_id, d in employees.items()
    if d.get("Department") == "IT"
}

print("Q40:", it_employees)
```
